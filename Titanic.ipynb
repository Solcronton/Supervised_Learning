```python
import numpy as np
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset("titanic")
```

```python
df
```

```python
df.columns
```

```python
df.info()
```

```python
df.drop(['deck', 'embark_town',
       'alive', 'class', 'who', 'adult_male'],axis=1,inplace=True)
```

```python
df.info()
```

```python
df['age'].fillna(df['age'].mean(), inplace=True)
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
df.head()
```

```python
df['sex']=le.fit_transform(df['sex'])
```

```python
df['embarked']=le.fit_transform(df['embarked']) #s=2 , c=1, Q=1
```

```python
df=df.astype(int)
```

```python
df
```

```python
x=df.drop('survived',axis=1)
y=df['survived']
```

# Logistic 

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()

model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report,root_mean_squared_error
r=root_mean_squared_error(y_test,y_pred)
```

```python
r
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
